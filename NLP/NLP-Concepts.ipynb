# ***Tokenization***

***Tokenization is the foundational process in NLP where raw text is broken down into smaller, manageable units called "tokens" (such as words, subwords, or characters). It acts as the crucial first step for language models and algorithms, converting unstructured text into a numerical format they can understand***

```python
corpus = """Tokenization is the foundational process in NLP where raw text is broken down into smaller, manageable units called "tokens".
            It's very important for NLP as it is required for performing various tasks. """
```

```python
import nltk
nltk.download('punkt_tab') #It's necessary to download this to use the tokenization functions in NLTK
```

```python
from nltk.tokenize import word_tokenize, sent_tokenize
Tokens = word_tokenize(corpus)
Documents = sent_tokenize(corpus)
```

```python
print("Senteces in the corpus are: ", Documents)
print("Tokens in the corpus are: ", Tokens)
```

```python
from nltk.tokenize import wordpunct_tokenize
PunctTokens = wordpunct_tokenize(corpus)
print("Tokens in the corpus with punctuatiions are: ", PunctTokens)
```

# ***Stemming***

***Stemming is a text preprocessing technique in Natural Language Processing (NLP) that reduces words to their root or base form by removing prefixes and suffixes. Its primary goal is to normalize text by grouping different variations of a word into a single, standardized root, even if that root is not a real dictionary word.***

*Stemming uses fixed, rule-based algorithms to mechanically chop off word endings without considering the word's context or part of speech.*

```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

* PorterStemmer

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()

for word in words:
    print(word + " : " + stemmer.stem(word))
```

```python
for word in Tokens:
    print(word + " : " + stemmer.stem(word))
```

* SnowballStemmer

```python
from nltk.stem import SnowballStemmer
SnowStemmer = SnowballStemmer(language = 'english')

for word in words:
    print(word + " : " + SnowStemmer.stem(word))
```

```python
for word in Tokens:
    print(word + " : " + SnowStemmer.stem(word))
```

* LancesterStemming

```python
"""
LanasterStemming is an aggressive stemming which removes more characters from the word than the other two stemmers.
It is not recommended to use this stemmer in production as it can lead to loss of information. It also let the user
set their own rules for stemming as LancasterStemmer() has two parameters.
"""

from nltk.stem import LancasterStemmer
LanceStemmer = LancasterStemmer()

for word in words:
    print(word + " : " + LanceStemmer.stem(word))
```

# ***Lemmatization***

***Lemmatization is an NLP text normalization technique that reduces inflected words to their meaningful base or dictionary form (called a lemma). Unlike stemming, which merely chops off prefixes and suffixes, lemmatization relies on morphological analysis and context (Part-of-Speech tagging) to produce valid, dictionary-recognized words.***

*It uses vocabulary, context, and part-of-speech rules to reduce words to their actual dictionary base (or lemma). It is more accurate but computationally heavier.*

```python
nltk.download('wordnet') #It's important to download this to use the lemmatization functions in NLTK
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

for word in words:
    print(word + " : " + lemmatizer.lemmatize(word, pos = 'v')) #pos is the part of speech tag which is used to lemmatize the word. It can be 'n' for noun, 'v' for verb, 'a' for adjective and 'r' for adverb.

#The pos decides how the will be treated as noun or verb or adjective or adverb
```

```python
for word in Tokens:
    print(word + " : " + lemmatizer.lemmatize(word, pos = 'v'))
```

#***StopWords***

***Stop words are high-frequency, low-meaning words (such as "the," "is," "at," and "and") that are frequently filtered out during natural language processing (NLP) and information retrieval tasks.***

*Removing Stopwords reduces vocabulary size helping algorithms to process the corpus faster.also it helps count-based methods (like Bag-of-Words or TF-IDF) to place more weight on content-bearing nouns and verbs. Removing them can hurt performance in tasks where grammar, sequence, or negation matters—such as sentiment analysis, where words like "not" or auxiliary verbs carry crucial context. Modern transformer models generally do not strip stop words because attention mechanisms learn which words matter.*

```python
from nltk.corpus import stopwords
nltk.download("stopwords")
```

```python
stopwords.words('english')
```

```python
para = """Stop words are high-frequency, low-meaning words (such as "the," "is," "at," and "and") that are frequently filtered out during natural language processing (NLP) and information retrieval tasks.
Removing Stopwords reduces vocabulary size helping algorithms to process the corpus faster.also it helps count-based methods (like Bag-of-Words or TF-IDF) to place more weight on content-bearing nouns and verbs.
Removing them can hurt performance in tasks where grammar, sequence, or negation matters—such as sentiment analysis, where words like "not" or auxiliary verbs carry crucial context.
Modern transformer models generally do not strip stop words because attention mechanisms learn which words matter."""
```

```python
para_sent = sent_tokenize(para)
```

```python
for i in range(len(para_sent)):
  words = nltk.word_tokenize(para_sent[i])
  word = [lemmatizer.lemmatize(word.lower()) for word in words if word not in set(stopwords.words('english'))]
  para_sent[i] = ' '.join(word)
```

# ***Parts-of-Speech (POS)***

***POS Tagging is the process in Natural Language Processing (NLP) where each word (token) in a text is assigned a grammatical category—such as a noun, verb, adjective, or adverb—based on both its definition and its surrounding context***

*Words can change parts of speech depending on how they are used. For example, in “The company will lead the project” versus “The pipe is made of lead,” POS tagging helps the system know whether "lead" is a verb or a noun.It acts as a foundational layer for parsing, machine translation, named entity recognition (NER), sentiment analysis, and question-answering systems*

```python
nltk.download('averaged_perceptron_tagger_eng')
## POS Tagging requires a list of words

for i in range(len(para_sent)):
  words = nltk.word_tokenize(para_sent[i])
  word = [lemmatizer.lemmatize(word.lower()) for word in words if word not in set(stopwords.words('english'))]
  tags = nltk.pos_tag(word)
  print("POS Tagged ->", tags)
```

# ***Named Entity Recognition(NER)***

***NER is a natural language processing (NLP) task that locates and classifies specific pieces of information—such as people, organizations, locations, dates, and monetary values—in unstructured text***

*It Uses models such as Conditional Random Fields (CRFs) or Support Vector Machines trained on labeled text features.It Uses neural architectures like BiLSTM-CRFs, Transformers (e.g., BERT), and instruction-tuned or structured-output large language models to capture rich context and few-shot or custom labels*

```python
Ner_sent = "On November 23, 2026, Tata Consultancy Services CEO K. Krithivasan announced in Mumbai that the company will invest ₹1,200 crore ($140 million) to expand its AI cloud platform, Ignio, across Kolkata and Hyderabad by March 2027, aiming to boost regional enterprise revenue by 18%."
```

```python
tag_elements = nltk.pos_tag(Ner_sent.split(' '))
```

```python
tag_elements
```

```python

from nltk import ne_chunk
nltk.download('maxent_ne_chunker')
nltk.download('words')

ne = nltk.ne_chunk(tag_elements)
print(ne)
```
