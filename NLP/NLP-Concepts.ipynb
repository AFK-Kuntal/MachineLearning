# ***Tokenization***

***Tokenization is the foundational process in NLP where raw text is broken down into smaller, manageable units called "tokens" (such as words, subwords, or characters). It acts as the crucial first step for language models and algorithms, converting unstructured text into a numerical format they can understand***

```python
corpus = """Tokenization is the foundational process in NLP where raw text is broken down into smaller, manageable units called "tokens".
            It's very important for NLP as it is required for performing various tasks. """
```

```python
import nltk
nltk.download('punkt_tab') #It's necessary to download this to use the tokenization functions in NLTK
```

```python
from nltk.tokenize import word_tokenize, sent_tokenize
Tokens = word_tokenize(corpus)
Documents = sent_tokenize(corpus)
```

```python
print("Senteces in the corpus are: ", Documents)
print("Tokens in the corpus are: ", Tokens)
```

```python
from nltk.tokenize import wordpunct_tokenize
PunctTokens = wordpunct_tokenize(corpus)
print("Tokens in the corpus with punctuatiions are: ", PunctTokens)
```

# ***Stemming***

***Stemming is a text preprocessing technique in Natural Language Processing (NLP) that reduces words to their root or base form by removing prefixes and suffixes. Its primary goal is to normalize text by grouping different variations of a word into a single, standardized root, even if that root is not a real dictionary word.***

*Stemming uses fixed, rule-based algorithms to mechanically chop off word endings without considering the word's context or part of speech.*

```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

* PorterStemmer

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()

for word in words:
    print(word + " : " + stemmer.stem(word))
```

```python
for word in Tokens:
    print(word + " : " + stemmer.stem(word))
```

* SnowballStemmer

```python
from nltk.stem import SnowballStemmer
SnowStemmer = SnowballStemmer(language = 'english')

for word in words:
    print(word + " : " + SnowStemmer.stem(word))
```

```python
for word in Tokens:
    print(word + " : " + SnowStemmer.stem(word))
```

* LancesterStemming

```python
"""
LanasterStemming is an aggressive stemming which removes more characters from the word than the other two stemmers.
It is not recommended to use this stemmer in production as it can lead to loss of information. It also let the user
set their own rules for stemming as LancasterStemmer() has two parameters.
"""

from nltk.stem import LancasterStemmer
LanceStemmer = LancasterStemmer()

for word in words:
    print(word + " : " + LanceStemmer.stem(word))
```

# ***Lemmatization***

***Lemmatization is an NLP text normalization technique that reduces inflected words to their meaningful base or dictionary form (called a lemma). Unlike stemming, which merely chops off prefixes and suffixes, lemmatization relies on morphological analysis and context (Part-of-Speech tagging) to produce valid, dictionary-recognized words.***

*It uses vocabulary, context, and part-of-speech rules to reduce words to their actual dictionary base (or lemma). It is more accurate but computationally heavier.*

```python
nltk.download('wordnet') #It's important to download this to use the lemmatization functions in NLTK
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

for word in words:
    print(word + " : " + lemmatizer.lemmatize(word, pos = 'v')) #pos is the part of speech tag which is used to lemmatize the word. It can be 'n' for noun, 'v' for verb, 'a' for adjective and 'r' for adverb.

#The pos decides how the will be treated as noun or verb or adjective or adverb
```

```python
for word in Tokens:
    print(word + " : " + lemmatizer.lemmatize(word, pos = 'v'))
```

#***StopWords***

***Stop words are high-frequency, low-meaning words (such as "the," "is," "at," and "and") that are frequently filtered out during natural language processing (NLP) and information retrieval tasks.***

*Removing Stopwords reduces vocabulary size helping algorithms to process the corpus faster.also it helps count-based methods (like Bag-of-Words or TF-IDF) to place more weight on content-bearing nouns and verbs. Removing them can hurt performance in tasks where grammar, sequence, or negation matters—such as sentiment analysis, where words like "not" or auxiliary verbs carry crucial context. Modern transformer models generally do not strip stop words because attention mechanisms learn which words matter.*

## ***One Hot Encoding***

*One-hot encoding in Natural Language Processing (NLP) converts each unique word or token in a vocabulary into a binary vector of length equal to the vocabulary size, with a 1 at the word's index and 0s everywhere else. For example, if a vocabulary has 3 words (["Kolkata", "Delhi", "Mumbai"]), "Kolkata" becomes [1, 0, 0], "Delhi" becomes [0, 1, 0], and "Mumbai" becomes [0, 0, 1].*

**Advantages**

---



• Simple and intuitive: Easy to understand and implement for basic text preprocessing or small vocabularies.

• No false ordinality: Unlike integer/label encoding, it does not assign arbitrary rankings or numerical order between distinct words.

• Useful building block: Works well for small categorical features alongside text or as an input representation before embedding layers and softmax outputs.

**Disadvantages**

---



• High dimensionality and sparsity: A vocabulary of 50,000 words produces 50,000-length vectors mostly filled with 0s, wasting memory and compute.

• No semantic meaning: Every pair of distinct word vectors is orthogonal (equally distant), so words like "good" and "great" have no built-in similarity.

• No context awareness: Polysemous words (words with multiple meanings) receive the same vector regardless of context.

• Out-of-vocabulary (OOV) issues: Unseen words during inference cannot be encoded without a predefined fallback or handling strategy.

```python
from sklearn.preprocessing import OneHotEncoder
import numpy as np

corpus = ["Kolkata is lively", "Delhi is busy"]
tokens = np.array(" ".join(corpus).split()).reshape(-1, 1)

encoder = OneHotEncoder(sparse_output=False, handle_unknown="ignore")
encoded = encoder.fit_transform(tokens)

print("Categories:", encoder.categories_[0])
print("Encoded shape:", encoded.shape)
print(encoded)
```

# ***Bag of Words***

*BOW is a foundational NLP technique that represents a document or sentence as a fixed-length vector showing how many times each vocabulary word appears, completely ignoring grammar and word order.*

### **Type**

---




*   Binary BOW : Stores 1 if a word appears in the document and 0 otherwise, ignoring how often it repeats. use when Short texts (like tweets or search queries) where repeated words add little extra signal.

*   Count Based BOW : Stores raw frequency counts (0, 1, 2, ...) of each word in a document. use when Standard baseline text classification when word repetition matters.

*   N-gram BOW : Counts sequences of n consecutive tokens (unigrams, bigrams n=2, trigrams n=3, or ranges like (1, 2)) instead of single words alone. Use when Phrases and negation patterns (e.g., "not good" or "Howrah Bridge").


### **How It Works (Example)**

---



Consider two sentences:
1. "Kolkata is lively and Kolkata is historic"
2. "Delhi is busy"

• Step 1 — Build Vocabulary: ["and", "busy", "Delhi", "historic", "is", "Kolkata", "lively"] (7 unique words)

• Step 2 — Count Word Frequencies:

	• Sentence 1: [1, 0, 0, 1, 2, 2, 1] ("Kolkata" and "is" appear twice)
	• Sentence 2: [0, 1, 1, 0, 1, 0, 0]


### **• Advantages:**

---


• Simple, fast, and effective for baseline text classification (like spam detection or sentiment analysis).

• Produces fixed-length vectors regardless of how long the input text is.

• Unlike One Hot Encoding it turns the whole sentence into one vector not every vector turns into different vectors







### **• Disadvantages:**

---



• Loses word order and context: "dog bites man" and "man bites dog" produce the exact same vector.

• High sparsity: Most entries are 0 for large vocabularies.

• Common words dominate: Frequent words like "is" or "the" get high counts even though they carry little meaning (which TF-IDF solves).


```python
from sklearn.feature_extraction.text import CountVectorizer

corpus = [
    "Kolkata metro is fast and fast",
    "Metro in Kolkata is crowded"
]

# 1. Binary BoW
binary_vec = CountVectorizer(binary=True)
bow_vec = binary_vec.fit_transform(corpus)

print("Binary BOW")
print("Vocabulary:", binary_vec.get_feature_names_out())
print("Binary:\n", bow_vec.toarray())

#2. Count BoW
count_vec = CountVectorizer()
bow_vec = count_vec.fit_transform(corpus)

print("Count BOW")
print("Vocabulary:", count_vec.get_feature_names_out())
print("Count:\n", bow_vec.toarray())

# 3. Bigram BoW (n=2)
bigram_vec = CountVectorizer(ngram_range=(2, 2))
bigram_bow_vec = bigram_vec.fit_transform(corpus)

print("N-gram BOW")
print("Bigrams:", bigram_vec.get_feature_names_out())
print("BOW \n", bigram_bow_vec.toarray())
```

# ***TF-IDF***

*TF-IDF (Term Frequency–Inverse Document Frequency) is an NLP weighting technique that scores how important a word is to a specific document relative to an entire collection (corpus), boosting rare descriptive words and penalizing common words like "is" or "the".*


### **How the Math Works**

---



TF-IDF multiplies two scores together for every word t in document d:
1. Term Frequency (TF): How often the word appears in the current document.

$$\text{TF}(t, d) = \frac{\text{Count of } t \text{ in } d}{\text{Total words in } d}$$



2. Inverse Document Frequency (IDF): How rare the word is across all N documents in the corpus.


$$\text{IDF}(t) = \log\left(\frac{N}{\text{Number of documents containing } t}\right)$$



3. Combined Score:

$$\text{TF-IDF}(t, d) = \text{TF}(t, d) \times \text{IDF}(t)$$


• High score: Word appears frequently in this document, but rarely in other documents (e.g., "tram" in an article about Kolkata).

• Zero / Low score: Word appears in almost every document (e.g., "the", "is"), making \(\log(N/N) = \log(1) = 0\).


### TF-IDF vs. Bag of Words (BoW)

---



| Feature | Bag of Words (Count) | TF-IDF |
| :--- | :--- | :--- |
| **Cell values** | Raw integers (`0, 1, 2, ...`) | Continuous decimal weights (`0.0` to `1.0`) |
| **Common words (`"is"`, `"and"`)** | Get high weights if repeated | Automatically downweighted |
| **Document length bias** | Longer docs naturally get higher counts | Normalized via L2 normalization |
| **Semantic similarity** | Still ignores word meaning and order | Still ignores word meaning and order |






### **• Advantages:**

---


• **Highlights key terms**: Boosts words that appear often in one document but rarely across the full collection, while downweighting common words.

• **Easy to calculate and inspect**: Uses straightforward arithmetic and produces transparent scores you can trace back to specific words.

• **Works as a fast baseline**: Requires no labeled data or pretraining, making it quick to test before moving to heavier models.

• **Language-independent**: Relies on word counts and document rarity rather than grammar rules, working across different languages.

• **Fixed-length vectors**: Turns texts of varying lengths into uniform numerical vectors for standard machine learning models.

### **• Disadvantages:**

---


• **No semantic or context awareness**: Treats words as independent tokens, missing synonyms, polysemy (words with multiple meanings), and overall intent.

• **Ignores word order**: Cannot distinguish phrases or sentence structure unless n-grams are added.

• **Out-of-vocabulary (OOV) limits**: Cannot score unseen words that were not in the training vocabulary.

• High-dimensional sparse matrices: Large vocabularies create big vectors full of zeros, increasing memory and compute overhead.

• **Corpus-dependent scores**: Changing the document collection changes IDF weights, and very rare terms or short texts can produce skewed scores..



```python
from sklearn.feature_extraction.text import TfidfVectorizer

corpus = [
    "Kolkata is famous for trams and sweets",
    "Delhi is famous for monuments",
    "Mumbai is famous for beaches"
]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(corpus)

print("Vocabulary:", vectorizer.get_feature_names_out())
print("TF-IDF Matrix:\n", tfidf_matrix.toarray().round(3))
```
